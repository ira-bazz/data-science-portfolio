При помощи линейных отображений:
- Уменьшить объект в два раза
- Повернуть на 130 градусов
- Отразить объект относительно прямой y=x (поможет матрица `[[0, 1, 0], [1, 0, 0], [0, 0, 1]]`)


````
### YOUR CODE HERE
import numpy as np
import matplotlib.pyplot as plt

A = [
    [-100, -100, 1],
    [-100, 100, 1],
    [100, 100, 1],
    [0, 0, 1],
    [100, -100, 1],
    [-100, -100, 1]
]
A = np.array(A)

# 1. Масштабирование
S = np.array([
    [0.5, 0, 0],
    [0, 0.5, 0],
    [0, 0, 1]
])

# 2. Поворот на 130 градусов
theta = np.deg2rad(130)
R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0, 0, 1]
])

# 3. Отражение относительно y=x
M = np.array([
    [0, 1, 0],
    [1, 0, 0],
    [0, 0, 1]
])

# Применяем все преобразования
A_new = A @ S.T @ R.T @ M.T

# исходный и преобразованный объект
plt.figure(figsize=(6,6))
plt.plot(A[:,0], A[:,1], label="Исходный")
plt.plot(A_new[:,0], A_new[:,1], label="Преобразованный")
plt.legend()
plt.ylim([-200,200])
plt.xlim([-200,200])
plt.grid(True)
plt.show()
````

### Задание 2

Найдите спектральное разложение матрицы:
```
m = np.array([[1, 2],
              [2, 3]])
print(m)
```



````
### YOUR CODE HERE
import numpy as np

m = np.array([[1, 2],
              [2, 3]])

# Собственные значения и векторы
eigenvalues, eigenvectors = np.linalg.eig(m)

print("Собственные значения:")
print(eigenvalues)

print("\nСобственные векторы (столбцы):")
print(eigenvectors)

# Проверка
Lambda = np.diag(eigenvalues)
V = eigenvectors
print("\nВосстановленная матрица:")
print(V @ Lambda @ np.linalg.inv(V))
````
