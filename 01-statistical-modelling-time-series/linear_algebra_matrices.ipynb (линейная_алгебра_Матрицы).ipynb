### Задание 1

 - Создайте матрицу 3 на 3, заполненную числами от 1 до 9
 - Прибавьте к ней матрицу с единицами на диагонале
 - Умножьте полученную матрицу на вектор (1, 2, 3)
 - Вычислить транспонированную матрицу $A^T$ от полученной в результате прибавления матрицы
 - Посчитайте обратную матрицу к полученной в результате прибавления матрице


````
### YOUR CODE HERE
import numpy as np

# 1) Матрица 3×3 с числами 1 - 9
A = np.arange(1, 10).reshape(3, 3)

# 2) Прибавляем единичную матрицу, единицы на диагонали
I = np.eye(3, dtype=int)
B = A + I                      # "полученная в результате прибавления матрица"

# 3) Умножаем B на вектор (1, 2, 3)
v = np.array([1, 2, 3])
prod = B.dot(v)

# 4) Транспонированная матрица B^T
BT = B.T

# 5) Обратная матрица к B
B_inv = np.linalg.inv(B)
````

### Задание 2

Найдите собственные значения и собственные вектора матрицы:
```
m = np.array([[2, 2],
              [1, 3]])
print(m)
```



````
### YOUR CODE HERE
import numpy as np

m = np.array([[2, 2],
              [1, 3]])

values, vectors = np.linalg.eig(m)

print("Собственные значения:\n", values)
print("\nСобственные векторы (по столбцам):\n", vectors)
````
